#### Characterstics of a List
- Ordered
- Mutable
- Hetrogeneous
- Can have duplicates
- are Dynamic
- can be nested
- items can be accessed
- can contain any kind of objects in python

Lists are ordered, means the order of items are different.

```python
L = [1,2,3]
L1 = [3,2,1]

L == L1
```

Here L is not equals to L1. Order matters.

```python
# List are mutable.
L = [1,2,3]
L[0] = 100
print(L)
```
