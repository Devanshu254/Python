#### Characterstics of a List
- Ordered
- Mutable
- Hetrogeneous
- Can have duplicates
- are Dynamic
- can be nested
- items can be accessed
- can contain any kind of objects in python

Lists are ordered, means the order of items are different.

```python
L = [1,2,3]
L1 = [3,2,1]

L == L1
```

Here L is not equals to L1. Order matters.

```python
# List are mutable.
L = [1,2,3]
L[0] = 100
print(L)
```

#### Creating a List
- Empty
- 1D
- 2D
- 3D
- Hetrogeneous
- Using type conversion

```python
# Empty List
print([])

# 1D -> Homogeneous.
print([1,2,3,4,5])

# 2D -> Heterogeneous list: contains integers and a nested list.
print([1,2,3,[4,5]])

# 3D List -> We can identify how many brackets are open by that we can understand whether the list is 3d or not.
# This list is homogenous.
print([
    [[1, 2],[3, 4]],
    [[5, 6],[6, 7]]
])

# Hetrogeneous list.
print([1, True, 5.6, 5+6j, 'Hello'])

# Using type conversion.
print(list('hello')) 
```
