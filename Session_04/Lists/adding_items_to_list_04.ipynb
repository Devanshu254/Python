### Adding Items to List
- append
- extend 
- insert

```python
# append
L = [1,2,3,4,5]
L.append(6)
print(L)
```

We can add multiple items to list with the help of extend() function.

```python
# extend
L = [1,2,3,4,5]
L.extend([6,7,8])
print(L)
```

If we try to add same thing with the help of append, it will always try to add a single item

```python
L = [1,2,3,4,5]
L.append([6,7,8])
print(L)
```
