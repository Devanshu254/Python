#### Accessing items.

```python
# Indexing.
L = [1,2,3,4,5]

# Positive Indexing: It goes from left to right and it starts from zero.
print(L[0])

# Negative Indexing.
print(L[-1])
```

How Indexing with work in 2D List.

```python
L = [1,2,3,[4,5]]

# Here we want to extract 4.
print(L[-1])
print(L[-1][-2])

# Same we can do with positive Indexing.
print(L[3][0])
```
