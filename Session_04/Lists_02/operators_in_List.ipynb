## Operations on List

- Arithmetic
- Membership
- Loop

```python
# Arithmetic
L1 = [1,2,3,4]
L2 = [5,6,7,8]

print(L1+L2)
```

```python
print(L1*3)
```

```python
# Membership operator.
# in/not in

L1 = [1,2,3,4,5]
L2 = [1,2,3,4,[5,6]]

print(5 in L1)
print(5 not in L1)
```

```python
print(5 in L2) # In 2D List the function behaves differently.
print([5,6] in L2)
```
