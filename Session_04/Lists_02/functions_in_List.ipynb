## Functions
- len/min/max/sorted.
- count.
- index.
- reverse.
- sort (vs sorted).
- copy.

## len/min/max/sorted.

```python
L = [2,1,5,7,0]
print(len(L))
print(min(L)) #If we have homogeneous data.
print(max(L))
print(sorted(L))
print(sorted(L, reverse=True))
```

## count
- Gives the frequency count in the list.

```python
L = [1,2,1,3,4,1,5]
L.count(1)
```

## index
- It will give index of a given item.

```python
L = [1,2,1,3,4,1,5]
L.index(2)
```

## reverse

```python
L = [1,2,1,3,4,1,5]
# It permanently reverses the list.
L.reverse()
print(L)
```

## sort (vs sorted)
- in sort list will be permanent change.
- in sorted, list will not be permanent change.
