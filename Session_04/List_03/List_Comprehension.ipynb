## List Comprehension

List Comprehension provides a concise way of creating lists.

newlist = [expression for item in iterable if condition == true]

### Advantages of List Comprehension
- More time efficient and space-efficient than loops.
- Require fewer lines of code.
- Transforms iterative statement into a formula.

```python
# Add 1 to 10 numbers in list
L = []
for i in range(1, 11):
    L.append(i)

print(L)
```

```python
# By Using List Comprehension.
L = [i for i in range(1, 11)]
print(L)
```

```python
# Scalar Multiplication of Vector.
v = [2,3,4]
s = -3

[s * i for i in v]
```

```python
# Add Squares
L = [1,2,3,4,5]
L = [i**2 for i in L]
print(L)
```

```python
# Print all the numbers which are divisible by 5 in range(1,50)
L = [i for i in range(1, 50) if i % 5 == 0]
print(L)
```

```python
# Find the languages which starts with P.
languages = ['java', 'python', 'php', 'c', 'c++', 'javascript']
[language for language in languages if language.startswith('p')]
```

```python
# Nested if with List Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple', 'kiwi', 'grapes', 'banana']

# add new list from my_fruits and items if the fruits exists in basket and also starts with 'a'
[fruit for fruit in my_fruits if fruit in basket and fruit.startswith('a')]
```
