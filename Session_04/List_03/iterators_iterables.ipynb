## What is an Iteration?
Iteration is a general term for taking each item of something, one after another. Any time you use a loop, explicit or implicit, to go over a group of items, that is iteration.

```python
# Example:
num = [1,2,3]

for i in num:
    print(i)
```

## Iterator
An Iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory.

```python
# Example:

L = [x for x in range(1,10000)]
# for i in L:
#     print(i*2)

import sys
print(sys.getsizeof(L) / 1024) # KB

x = range(1, 10000) # Internally it's calling an iterator that's the reason why the memory is not getting increased.
print(sys.getsizeof(x) / 1024) # KB
```

## Iterable
Iterable is an object, which one can iterate over.
It generates an iterator when passed to iter() method.

```python
# Example
L = [1,2,3]
print(type(L)) # <class 'list'>

# L is an iterable.
print(type(iter(L))) # <class 'list_iterator'>
```

## Points to remember
- Every iterator is iterable.
- Not all iterables are iterator.

## Trick
- Every iterable has iter function.
- Every iterator has both iter function as well as next function.

```python
# Either we will be able to run for() loop over a data type in order to know whether it's iterable.
a = 2
for i in a:
    print(i) # TypeError: 'int' object is not iterable
```
