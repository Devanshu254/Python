## Tuple
A tuple in Python is similar to list. The difference between the two is that we can not change the elements in tuple once it's assigned whereas we can change the elements of list.
In short, a tuple is an immutable list. A tuple can not be changed in any way once it's created.

### Characterstics
- Ordered
- Immutable
- Allows duplicate

<img src="images/list_tuple_com.png" width="70%"> <br>
Immutability is the fundamental difference. However, because tuples are immutable, they are more memory-efficient, slightly faster, hashable (can be used as dictionary keys), and are typically used for fixed data, whereas lists are used for data that needs to change.
