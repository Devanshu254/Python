## Static vs Dynamic Typing.
- In Statically typed language, the type of variable is fixed before program runs.
- The compiler already knows what type each variable is. <br>
Like in C++ <br>
int age = 26; <br>
here, age is an integer, you can not later store string in it.

- In a dynamically typed language, you don't declare the type.
- Python figures out the type automatically while the program is running. <br>
age = 26

## Static vs Dynamic Binding.
- When we call a function, computer decides which function to execute this process is known as binding.
- Suppose that we have a class.

```python
class Animal:
    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):
        print("Bark")
```

```python
animal = Dog()
animal.sound()
# Now Computer will think, which sound function is to call here?
```

In static binding, the compiler decides before the program runs which function will be called.

This is also called:

Early Binding
Compile-time Binding

Dynamic Binding (Late Binding)

In dynamic binding, the decision is made while the program is running.

This happens because the actual object might not be known until runtime.

Also called:

Late Binding
Runtime Binding
