## Type Casting
- Converting one kind of data type into another is known as type casting.

```python
a = 10
print(type(a))     # int

a = float(a)
print(a)           # 10.0
print(type(a))     # float
```

## There are mainly two types of Type casting in python.
- Implicit type casting.
- Explicit type casting.

### Implicit Type Casting
- Python converts automatically.

```python
a = 10       # int
b = 2.5      # float

c = a + b

print(c)        # 12.5
print(type(c))  # float

# Here python automatically converted 10 into 10.0
```

## Explicit Type Casting
- Programmer convert data type manually.
- To do this python's built in functions can be used.

```python
int()
float()
str()
bool()
```

```python
# Example
a = "10"

b = int(a)

print(b)        # 10
print(type(b))  # int
```

```python
# Common conversions

# String → Integer
a = int("10")

# Integer → Float
b = float(10)

# Integer → String
c = str(10)

# Integer → Boolean
d = bool(10)

# String → Boolean
e = bool("hello")
```
