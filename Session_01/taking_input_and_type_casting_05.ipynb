## Type Casting
- Converting one kind of data type into another is known as type casting.

```python
a = 10
print(type(a))     # int

a = float(a)
print(a)           # 10.0
print(type(a))     # float
```

## There are mainly two types of Type casting in python.
- Implicit type casting.
- Explicit type casting.

### Implicit Type Casting
- Python converts automatically.

```python
a = 10       # int
b = 2.5      # float

c = a + b

print(c)        # 12.5
print(type(c))  # float
```
