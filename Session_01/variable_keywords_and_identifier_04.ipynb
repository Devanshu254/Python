## Keyword
- A keyword in Python is a reserved word that has a predefined meaning and special purpose in the language, and it cannot be used as an identifier (such as a variable name or function name).

- if, else, for, while, def, class, return, True, False, None

## Identifier.
- An identifier in Python is a name used to identify a variable, function, class, module, or other object in a program.

```python
name = "Devanshu"
age = 22
# name → identifier
# age → identifier
```
